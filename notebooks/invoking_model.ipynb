```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model(
    model = "gpt-5-nano",
    use_responses_api = True,
    temperature = 1.0,    
)
```

```python
response = model.invoke("What is the capital of Moon?")
response
```
